 1 : What is Data Loading in Power BI and why is it considered the first step of analysis?

 Sol :       Data Loading in Power BI means importing data from different sources such as Excel, CSV files, SQL databases, websites, or cloud services into Power BI.

It is considered the first step of analysis because:

- Data must be available in Power BI before any cleaning, transformation, visualization, or reporting can be done.
- It connects raw data with analytical tools.
- Proper loading ensures accurate and efficient analysis.

3 : What is a Fact Table and a Dimension Table? Give examples from the dataset.

Sol :      

A Fact Table contains measurable numerical data used for analysis.

Examples from the dataset:

- Confirmed Cases
- Deaths
- Recovered Cases

A Dimension Table contains descriptive information related to facts.

Examples:

- Country Name
- State Name
- Date

4 : Why is Star Schema preferred over Snowflake Schema in Power BI?

Sol :       
Star Schema is preferred because:

i . Better Performance
- Fewer joins improve query speed.

ii . Simpler Structure
- Easier to understand and manage.

iii . Faster Report Generation
- Optimized for Power BI visuals and DAX calculations.

iv . Improved Usability
- Business users can work more easily with the model.
